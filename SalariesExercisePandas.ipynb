# SF Salaries Exercise - Solutions

Welcome to a quick exercise for you to practice your pandas knowledge. Dataset used [SF Salaries Dataset](https://www.kaggle.com/kaggle/sf-salaries). Just follow along and complete the tasks outlined in bold below.

** Import pandas as pd.**

```python
import pandas as pd
```

** Read Salaries.csv as a dataframe called sal.** 

If you are using CoLabs, you may need to check the documentation in how to load a CSV file.

```python
cols = ['TotalPayBenefits', 'OvertimePay', 'BasePay', 'OtherPay', 'Benefits']
na_values={c: ['Not Provided'] for c in cols}
sal = pd.read_csv('Salaries.csv', na_values=na_values)
```

** Check the head of the DataFrame. **

```python
sal.head()
```

** Use the .info() method to find out how many entries there are.**

```python
sal.info()
```

**What is the average BasePay ?**

```python
float(sal['BasePay'].mean())
```

** What is the highest amount of OvertimePay in the dataset ? **

```python
float(sal['OvertimePay'].max())
```

** What is the job title of  JOSEPH DRISCOLL ? Note: Use all caps, otherwise you may get an answer that doesn't match up (there is also a lowercase Joseph Driscoll). **

```python
sal[sal['EmployeeName'] == 'JOSEPH DRISCOLL']['JobTitle']
```

** How much does JOSEPH DRISCOLL make (including benefits)? **

```python
sal[sal['EmployeeName'] == 'JOSEPH DRISCOLL']['TotalPayBenefits']
```

** What is the name of highest paid person (including benefits)?**

```python
sal[
    sal['TotalPayBenefits']
    == sal['TotalPayBenefits'].max()
]
```

```python
sal[
    sal['TotalPayBenefits']
    == sal['TotalPayBenefits'].max()
]['EmployeeName']
```

** What is the name of lowest paid person (including benefits)? Do you notice something strange about how much he or she is paid?**

```python
sal[
    sal['TotalPayBenefits']
    == sal['TotalPayBenefits'].min()
]

# Joe Lopez does not receive any pay, and appears to owe money.
```

```python
sal[
    sal['TotalPayBenefits']
    == sal['TotalPayBenefits'].min()
]['EmployeeName']
```

** What was the average (mean) BasePay of all employees per year? (2011-2014) ? **

```python
sal.groupby('Year', as_index=False)['BasePay'].mean().round(6)
```

** How many unique job titles are there? **

```python
sal['JobTitle'].nunique()
```

** What are the top 5 most common jobs? **

```python
sal['JobTitle'].value_counts().head(5)
```

** How many Job Titles were represented by only one person in 2013? (e.g. Job Titles with only one occurence in 2013?) **

```python
int((sal[sal['Year'] == 2013]['JobTitle'].value_counts() == 1).sum())
```

** How many people have the word Chief in their job title? (This is pretty tricky) **

```python
sal[sal['JobTitle'].str.contains('Chief', case=False)]
```

```python
int(sal['JobTitle'].str.contains('Chief', case=False).sum())
```

** Optional: Is there a correlation between length of the Job Title string and Salary? **

```python
sal['title_len'] = sal['JobTitle'].str.len()
```

```python
sal[['title_len', 'TotalPayBenefits']].corr().round(6)
```
