## 8. From the category ‘news’, display the first words with its correspondent POS tag. 

```python
news_words = brown.tagged_words(categories='news')
news_words[:10]
```

```python
nouns = [(word, tag) for word, tag in news_words if tag.startswith(('NN', 'NP'))]
nouns[:10]
```

## 9. Build frequency distribution for Nouns and display top 5 occurring nouns 

```python
fdist = FreqDist(word for word, tag in nouns)
fdist.most_common(5)
```
